# Automação de Sistemas e Processos com Python

### Desafio Cotidiano:

Todos os dias, o nosso sistema atualiza as vendas do dia anterior.
A função desse projeto é enviar um e-mail para a diretoria da empresa X, assim que começar a trabalhar, com o faturamento e a quantidade de produtos vendidos no dia anterior

E-mail da diretoria: seugmail+diretoria@gmail.com<br>
Local onde o sistema disponibiliza as vendas do dia anterior: https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga

Para resolver isso, é utilizado o pyautogui, uma biblioteca de automação de comandos do mouse e do teclado


Referência do pyautogui: https://pyautogui.readthedocs.io/en/latest/quickstart.html

```python
#Passo a passo: Como você faria manualmente
#Passo 1: Abrir o site da empresa
#Passo 2: Navegar no sistema para encontrar base de dados
#Passo 3: Baixar arquivo
#Passo 4: Calcular indicadores - faturamento e quantidade de produtos vendidos
#Passo 5: Enviar informações por e-mail

#Bibliotecas utilizadas:
# pyautogui - automações; pandas - dados; openpyxl - excel; time - quanto tempo executa a função no meio dos comandos

import pyautogui
import time

link = "https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga"

pyautogui.PAUSE = 1
pyautogui.press("win")
pyautogui.write("chrome")
time.sleep(2)
pyautogui.press("enter")
time.sleep(5)
pyautogui.write(link)
pyautogui.press("enter")

time.sleep(10)
pyautogui.click(x=397, y=382, clicks = 2)
time.sleep(3)
pyautogui.click(x=1806, y=379)
pyautogui.click(x=1569, y=488)




```

### Leitura do arquivo baixado para pegar os indicadores

- Faturamento
- Quantidade de Produtos

```python
import pandas

caminho = r"C:\Users\barba\Downloads\Vendas - Dez.xlsx"
tabela = pandas.read_excel(caminho) #ler a base de dados

display(tabela) #utilizar display em vez de print para melhor visualização dos dados

faturamento = tabela["Valor Final"].sum()
qtde_produtos = tabela["Quantidade"].sum()

print(faturamento)
print(qtde_produtos)
```

### Vamos agora enviar um e-mail pelo gmail

```python
import pyperclip

time.sleep(10)
pyautogui.hotkey("ctrl", "t") #abrir nova aba

pyautogui.write("http://mail.google.com/")
pyautogui.press("enter")
time.sleep(10)
pyautogui.click(x=91, y=237) #conta já logada no gmail. Click para escrever novo e-mail
time.sleep(5)
pyautogui.write("emailteste@gmail.com")
pyautogui.press("tab")
pyautogui.press("tab")
pyperclip.copy("Relatório de Vendas")
pyautogui.hotkey("ctrl", "v")
pyautogui.press("tab")

texto = f"""
Prezados,

Segue Relatório de Vendas de hoje.

Faturamento: R${faturamento:,.2f}
Quantidade de produtos vendidos: {qtde_produtos:,}

Qualquer dúvida estou a disposição.

Atenciosamente,
Barbara Lascosck
"""
pyperclip.copy(texto)
pyautogui.hotkey("ctrl", "v")
pyautogui.click(x=1182, y=983)
```
